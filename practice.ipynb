## Practice With Real World Data

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("=== Zomato sales analysis ===")

  
```

```python
#Total sales per year
total_sales_per_year = np.sum(sales_data[:, 1:], axis=0)
print("Total sales per year:", total_sales_per_year)

#Minimum and maximum sales per restaurant
min_sales_per_restaurant = np.min(sales_data[:, 1:], axis=1)
max_sales_per_restaurant = np.max(sales_data[:, 1:], axis=1)
print("Minimum sales per restaurant:", min_sales_per_restaurant)
print("Maximum sales per restaurant:", max_sales_per_restaurant)

#Maximum Sales per year
max_sales_per_year = np.max(sales_data[:, 1:], axis=0)
print("Maximum sales per year:", max_sales_per_year)

#Average sales per restaurant
average_sales_per_restaurant = np.mean(sales_data[:, 1:], axis=1)
print("Average sales per restaurant:", average_sales_per_restaurant)

#Cumulative sales across all restaurants 
cumulative_sales = np.cumsum(sales_data[:, 1:], axis=1)
print("Cumulative sales across all restaurants:", cumulative_sales)

```

## General Questions

```python
# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

#### Get the shape of the matrix.

```python
print(data.shape)
```

#### Find the average age of students.

```python
avgAge =np.mean(data[:,0])
print("Average Age of Students:", avgAge)
```

#### Extract Math marks of all students.

```python
mathmarks = np.array(data[:,1])
print(mathmarks)
```

#### Find the highest Science mark.

```python
highsci = np.max(data[:,2])
print("Highest Science Marks:", highsci)
```

#### Get details of the student who scored more than 90 in Math.

```python
moreThan90 = data[data[:,1] > 90]
print(data[:,1]>90)
print("Students with Math Marks greater than 90:", moreThan90)
```

#### Increase Math marks of all students by 5.

```python
newMathMarks = np.array(data[:,1])+5
print("New Math Marks after adding 5:", newMathMarks)
```

#### Find how many students are younger than 19.

```python
syoungerThan19 = data[data[:,0]<19]
print("Students younger than 19:", len(syoungerThan19))
```

#### Calculate the average marks in each subject (column-wise mean).

```python
aVgMarks = np.mean(data[:,1:3], axis=0)
print("Average Marks in each subject:", aVgMarks)
```

#### Get data of students who scored at least 80 in both subjects.

```python
sb80 = data[(data[:,1] >= 80) & (data[:,2] >= 80)]
print("Students with at least 80 marks in both subjects:", sb80)
```

#### Replace all Science marks < 75 with 0.

```python
 data[:,2][data[:,2] < 75] = 0
 print("Science marks less than 75 set to 0:", data)
```
